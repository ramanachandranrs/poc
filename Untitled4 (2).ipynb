```python
# =============================================================================
#  AUTOMOTIVE SERVICE & INVENTORY — DATA VALIDATION + CORRELATION ANALYSIS
#  Ready to paste into a single Google Colab cell.
#  Files required (upload via the Colab sidebar or Files > Upload):
#    • daily_demand_fact.csv
#    • job_card_header.csv
#    • job_card_line_items.csv
# =============================================================================

# ── 0. IMPORTS ────────────────────────────────────────────────────────────────
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
from google.colab import files

# ── 1. FILE UPLOAD  ──────────────────────────────────────────────────────────
print("=" * 65)
print("  STEP 1 — UPLOAD YOUR CSV FILES")
print("  Select all three files at once in the dialog that opens.")
print("=" * 65)

uploaded = files.upload()   # opens the file-picker dialog

# ── 2. LOAD DATA ─────────────────────────────────────────────────────────────
import io

demand   = pd.read_csv(io.BytesIO(uploaded["daily_demand_fact.csv"]))
jc_hdr   = pd.read_csv(io.BytesIO(uploaded["job_card_header.csv"]))
jc_lines = pd.read_csv(io.BytesIO(uploaded["job_card_line_items.csv"]))

# Parse date columns upfront — keeps downstream checks tidy
demand["Date"]        = pd.to_datetime(demand["Date"],        errors="coerce")
jc_hdr["Date_In"]     = pd.to_datetime(jc_hdr["Date_In"],     errors="coerce")

print(f"\n  Loaded daily_demand_fact  : {demand.shape[0]:,} rows × {demand.shape[1]} cols")
print(f"  Loaded job_card_header    : {jc_hdr.shape[0]:,} rows × {jc_hdr.shape[1]} cols")
print(f"  Loaded job_card_line_items: {jc_lines.shape[0]:,} rows × {jc_lines.shape[1]} cols")

# ── 3. HELPER ─────────────────────────────────────────────────────────────────
PASS = "✅  PASS"
FAIL = "❌  FAIL"
WARN = "⚠️   WARN"

def section(title):
    print(f"\n{'=' * 65}")
    print(f"  CHECK {title}")
    print(f"{'=' * 65}")

# =============================================================================
#  CHECK 1 — MISSING / NULL VALUES IN CRITICAL FIELDS
# =============================================================================
section("1 — Missing / Null Values in Critical Fields")

critical_cols = ["Demand_ID", "Date", "Part_Number",
                 "Demand_Qty", "Sales_Value", "Unit_Price"]

missing = demand[critical_cols].isnull().sum()
total   = len(demand)

for col, n in missing.items():
    pct    = n / total * 100
    status = PASS if n == 0 else FAIL
    print(f"  {status}  {col:<25} nulls = {n:,}  ({pct:.2f} %)")

null_dates = demand["Date"].isnull().sum()
if null_dates:
    print(f"\n  {FAIL}  {null_dates:,} rows had unparseable Date values (set to NaT).")

# =============================================================================
#  CHECK 2 — NO NEGATIVE STOCK (On_Hand_Qty)
# =============================================================================
section("2 — No Negative Stock Levels (On_Hand_Qty)")

neg_stock = demand[demand["On_Hand_Qty"] < 0]
if neg_stock.empty:
    print(f"  {PASS}  All On_Hand_Qty values are ≥ 0.")
else:
    print(f"  {FAIL}  {len(neg_stock):,} rows have negative On_Hand_Qty.")
    print(neg_stock[["Demand_ID", "Part_Number", "On_Hand_Qty"]].head(10).to_string(index=False))

# Also flag negatives in Demand_Qty (sanity check)
neg_demand = demand[demand["Demand_Qty"] < 0]
if neg_demand.empty:
    print(f"  {PASS}  All Demand_Qty values are ≥ 0.")
else:
    print(f"  {WARN}  {len(neg_demand):,} rows have negative Demand_Qty.")

# =============================================================================
#  CHECK 3 — SALES_VALUE == DEMAND_QTY × UNIT_PRICE  (float tolerance)
# =============================================================================
section("3 — Sales_Value = Demand_Qty × Unit_Price")

TOLERANCE = 0.01   # ₹ / $ tolerance per row

demand["_computed_sales"] = demand["Demand_Qty"] * demand["Unit_Price"]
demand["_sales_diff"]     = (demand["Sales_Value"] - demand["_computed_sales"]).abs()

mismatch = demand[demand["_sales_diff"] > TOLERANCE]

if mismatch.empty:
    print(f"  {PASS}  All {total:,} rows satisfy Sales_Value = Demand_Qty × Unit_Price"
          f" (tolerance ± {TOLERANCE}).")
else:
    print(f"  {FAIL}  {len(mismatch):,} rows have Sales_Value inconsistency.")
    print(mismatch[["Demand_ID", "Part_Number", "Demand_Qty",
                    "Unit_Price", "Sales_Value",
                    "_computed_sales", "_sales_diff"]].head(10).to_string(index=False))

# Drop temp columns
demand.drop(columns=["_computed_sales", "_sales_diff"], inplace=True)

# =============================================================================
#  CHECK 4 — REORDER LOGIC: SOQ_Suggested_Qty should be 0
#             when On_Hand_Qty > Reorder_Point
# =============================================================================
section("4 — Reorder Logic (no SOQ when On_Hand_Qty > Reorder_Point)")

logic_violation = demand[
    (demand["On_Hand_Qty"] > demand["Reorder_Point"]) &
    (demand["SOQ_Suggested_Qty"] > 0)
]

if logic_violation.empty:
    print(f"  {PASS}  No automated orders placed when stock was above Reorder Point.")
else:
    print(f"  {FAIL}  {len(logic_violation):,} rows triggered an SOQ despite "
          f"On_Hand_Qty > Reorder_Point.")
    print(logic_violation[["Demand_ID", "Part_Number",
                            "On_Hand_Qty", "Reorder_Point",
                            "SOQ_Suggested_Qty"]].head(10).to_string(index=False))

# =============================================================================
#  CHECK 5 — REFERENTIAL INTEGRITY: Job-Card Line Items → Header
# =============================================================================
section("5 — Referential Integrity (Line Items → Header)")

valid_jc_numbers = set(jc_hdr["Job_Card_Number"].dropna().unique())
line_jc_numbers  = set(jc_lines["Job_Card_Number"].dropna().unique())

orphan_keys = line_jc_numbers - valid_jc_numbers

if not orphan_keys:
    print(f"  {PASS}  All {len(line_jc_numbers):,} unique Job_Card_Numbers in line items "
          f"exist in the header table.")
else:
    orphan_rows = jc_lines[jc_lines["Job_Card_Number"].isin(orphan_keys)]
    print(f"  {FAIL}  {len(orphan_keys):,} orphan Job_Card_Numbers "
          f"({len(orphan_rows):,} rows) have no matching header record.")
    print(f"         Sample orphan keys: {list(orphan_keys)[:5]}")

# Reverse: headers with no line items (informational only)
headers_no_lines = valid_jc_numbers - line_jc_numbers
if headers_no_lines:
    print(f"\n  {WARN}  {len(headers_no_lines):,} Job_Card_Numbers in the header have "
          f"zero line items (may be valid for service-only jobs).")

# =============================================================================
#  CHECK 6 — DATE RANGE
# =============================================================================
section("6 — Operational Date Range")

d_min = demand["Date"].min()
d_max = demand["Date"].max()
jc_min = jc_hdr["Date_In"].min()
jc_max = jc_hdr["Date_In"].max()

print(f"  daily_demand_fact  → Date    : {d_min.date()}  →  {d_max.date()}"
      f"  ({(d_max - d_min).days + 1:,} calendar days)")
print(f"  job_card_header    → Date_In : {jc_min.date()}  →  {jc_max.date()}"
      f"  ({(jc_max - jc_min).days + 1:,} calendar days)")

# Flag future dates
future_demand = demand[demand["Date"] > pd.Timestamp.today()]
future_jc     = jc_hdr[jc_hdr["Date_In"] > pd.Timestamp.today()]
if future_demand.empty and future_jc.empty:
    print(f"  {PASS}  No future-dated records found.")
else:
    if not future_demand.empty:
        print(f"  {WARN}  {len(future_demand):,} demand rows have future dates.")
    if not future_jc.empty:
        print(f"  {WARN}  {len(future_jc):,} job-card rows have future dates.")

# =============================================================================
#  CHECK 7 — OUTLIER DETECTION  (IQR × 1.5  and  IQR × 3.0)
# =============================================================================
section("7 — Outlier Detection via IQR (Demand_Qty & Sales_Value)")

def iqr_outliers(series, label, multipliers=(1.5, 3.0)):
    """Return a dict of outlier DataFrames keyed by multiplier."""
    q1, q3 = series.quantile(0.25), series.quantile(0.75)
    iqr    = q3 - q1
    print(f"\n  [{label}]  Q1={q1:.2f}  Q3={q3:.2f}  IQR={iqr:.2f}  "
          f"Median={series.median():.2f}  Mean={series.mean():.2f}")
    results = {}
    for m in multipliers:
        lower  = q1 - m * iqr
        upper  = q3 + m * iqr
        flags  = (series < lower) | (series > upper)
        n      = flags.sum()
        status = PASS if n == 0 else WARN
        print(f"    {status}  ×{m} IQR  fence=[{lower:.2f}, {upper:.2f}]  "
              f"outliers = {n:,}  ({n/len(series)*100:.2f} %)")
        results[m] = demand[flags]
    return results

dq_outliers  = iqr_outliers(demand["Demand_Qty"],   "Demand_Qty")
sv_outliers  = iqr_outliers(demand["Sales_Value"],  "Sales_Value")

# ─── SUMMARY TABLE ────────────────────────────────────────────────────────────
print("\n")
print("=" * 65)
print("  VALIDATION SUMMARY")
print("=" * 65)
checks = [
    ("1 · Null critical fields",      missing.sum() == 0),
    ("2 · No negative stock",         neg_stock.empty),
    ("3 · Sales_Value consistency",   mismatch.empty),
    ("4 · Reorder logic",             logic_violation.empty),
    ("5 · Referential integrity",     not orphan_keys),
    ("6 · Date range (see above)",    True),          # always informational
    ("7 · Outliers flagged (IQR×1.5)",
         len(dq_outliers[1.5]) == 0 and len(sv_outliers[1.5]) == 0),
]
for desc, ok in checks:
    print(f"  {'✅' if ok else '❌'}  {desc}")

# =============================================================================
#  PLOT A — CORRELATION MATRIX HEATMAP
# =============================================================================
print("\n\n  Generating plots …")

numeric_cols = ["Demand_Qty", "On_Hand_Qty", "Reorder_Point",
                "SOQ_Suggested_Qty", "Unit_Price", "Sales_Value"]

corr = demand[numeric_cols].corr()

fig_corr, ax_corr = plt.subplots(figsize=(9, 7))
sns.heatmap(
    corr,
    annot=True,
    fmt=".2f",
    cmap="RdYlGn",
    center=0,
    linewidths=0.5,
    linecolor="white",
    square=True,
    ax=ax_corr,
    annot_kws={"size": 11},
)
ax_corr.set_title("Correlation Matrix — Demand & Inventory Numerics",
                  fontsize=14, fontweight="bold", pad=14)
plt.tight_layout()
plt.savefig("correlation_heatmap.png", dpi=150, bbox_inches="tight")
plt.show()
print("  Saved → correlation_heatmap.png")

# =============================================================================
#  PLOT B — BOXPLOTS FOR OUTLIER VISUALISATION
# =============================================================================
fig_box = plt.figure(figsize=(16, 10))
fig_box.suptitle("Outlier Visualisation — Demand Qty & Sales Value",
                 fontsize=15, fontweight="bold", y=1.01)

gs = gridspec.GridSpec(2, 3, figure=fig_box, hspace=0.45, wspace=0.35)

# ── Row 0: Demand_Qty ────────────────────────────────────────────────────────
ax0 = fig_box.add_subplot(gs[0, 0])
sns.boxplot(y=demand["Demand_Qty"], color="#4C9BE8", flierprops=dict(
    marker="o", markerfacecolor="red", markersize=4, alpha=0.5), ax=ax0)
ax0.set_title("Demand_Qty\nFull Distribution", fontsize=11)
ax0.set_ylabel("Demand_Qty")

ax1 = fig_box.add_subplot(gs[0, 1])
flag_15_dq = (
    (demand["Demand_Qty"] < demand["Demand_Qty"].quantile(0.25)
                            - 1.5 * (demand["Demand_Qty"].quantile(0.75)
                                     - demand["Demand_Qty"].quantile(0.25))) |
    (demand["Demand_Qty"] > demand["Demand_Qty"].quantile(0.75)
                            + 1.5 * (demand["Demand_Qty"].quantile(0.75)
                                     - demand["Demand_Qty"].quantile(0.25)))
)
colors_dq = np.where(flag_15_dq, "red", "#4C9BE8")
ax1.scatter(range(len(demand)), demand["Demand_Qty"],
            c=colors_dq, s=6, alpha=0.5)
ax1.set_title("Demand_Qty\n× 1.5 IQR Outliers (red)", fontsize=11)
ax1.set_xlabel("Row index")
ax1.set_ylabel("Demand_Qty")

ax2 = fig_box.add_subplot(gs[0, 2])
top_dq = demand.nlargest(20, "Demand_Qty")
sns.barplot(x=top_dq["Demand_Qty"].values,
            y=top_dq["Part_Number"].astype(str).values,
            palette="Reds_r", ax=ax2)
ax2.set_title("Top 20 Demand_Qty\nby Part Number", fontsize=11)
ax2.set_xlabel("Demand_Qty")
ax2.set_ylabel("Part_Number")

# ── Row 1: Sales_Value ───────────────────────────────────────────────────────
ax3 = fig_box.add_subplot(gs[1, 0])
sns.boxplot(y=demand["Sales_Value"], color="#F4A261", flierprops=dict(
    marker="o", markerfacecolor="red", markersize=4, alpha=0.5), ax=ax3)
ax3.set_title("Sales_Value\nFull Distribution", fontsize=11)
ax3.set_ylabel("Sales_Value")

ax4 = fig_box.add_subplot(gs[1, 1])
q1_sv = demand["Sales_Value"].quantile(0.25)
q3_sv = demand["Sales_Value"].quantile(0.75)
iqr_sv = q3_sv - q1_sv
flag_15_sv = (
    (demand["Sales_Value"] < q1_sv - 1.5 * iqr_sv) |
    (demand["Sales_Value"] > q3_sv + 1.5 * iqr_sv)
)
colors_sv = np.where(flag_15_sv, "red", "#F4A261")
ax4.scatter(range(len(demand)), demand["Sales_Value"],
            c=colors_sv, s=6, alpha=0.5)
ax4.set_title("Sales_Value\n× 1.5 IQR Outliers (red)", fontsize=11)
ax4.set_xlabel("Row index")
ax4.set_ylabel("Sales_Value")

ax5 = fig_box.add_subplot(gs[1, 2])
demand["_month"] = demand["Date"].dt.to_period("M").astype(str)
monthly_sales = (demand.groupby("_month", sort=True)["Sales_Value"]
                 .sum().reset_index())
sns.lineplot(data=monthly_sales, x="_month", y="Sales_Value",
             marker="o", color="#E76F51", ax=ax5)
ax5.set_title("Monthly Total Sales_Value\n(trend check)", fontsize=11)
ax5.set_xlabel("Month")
ax5.set_ylabel("Total Sales_Value")
ax5.tick_params(axis="x", rotation=45)
demand.drop(columns=["_month"], inplace=True)

plt.tight_layout()
plt.savefig("outlier_boxplots.png", dpi=150, bbox_inches="tight")
plt.show()
print("  Saved → outlier_boxplots.png")

# =============================================================================
#  DONE
# =============================================================================
print("\n" + "=" * 65)
print("  ALL CHECKS COMPLETE. Both plots saved to the Colab file system.")
print("  Download them via Files (left sidebar) → right-click → Download.")
print("=" * 65)
```

```python
# =============================================================
#  MODULE 1 — DEMAND FORECASTING
#  Models: Prophet (baseline) + XGBoost (feature-rich)
#  Install once: pip install prophet xgboost scikit-learn
# =============================================================

# ── 0. INSTALL & IMPORTS ─────────────────────────────────────
!pip install prophet xgboost scikit-learn --quiet

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

from prophet import Prophet
from xgboost import XGBRegressor
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import (mean_absolute_error,
                             mean_squared_error, r2_score)

# ── 1. LOAD CLEAN DATA ────────────────────────────────────────
demand_clean = pd.read_csv("demand_clean.csv",
                           parse_dates=["Date"])

print(f"Loaded: {len(demand_clean):,} rows")
print(f"Date range: {demand_clean['Date'].min().date()} "
      f"→ {demand_clean['Date'].max().date()}")

# ── 2. AGGREGATE TO DAILY TOTAL DEMAND ───────────────────────
# Prophet works on an aggregated time series.
# We forecast total daily demand first, then drill into
# category/part level using XGBoost.

daily = (demand_clean
         .groupby("Date")
         .agg(
             total_demand   = ("Demand_Qty",        "sum"),
             avg_price      = ("Unit_Price",         "mean"),
             promo_active   = ("Promotion_Flag",     "max"),
             recall_active  = ("Recall_Flag",        "max"),
             new_vehicle    = ("New_Vehicle_Flag",   "max"),
             stockout_count = ("Stockout_Flag",      "sum"),
         )
         .reset_index()
         .sort_values("Date"))

print(f"\nDaily aggregated: {len(daily)} days")

# ── 3. FEATURE ENGINEERING ────────────────────────────────────
def engineer_features(df):
    df = df.copy()

    # Calendar features
    df["day_of_week"]    = df["Date"].dt.dayofweek          # 0=Mon
    df["day_of_month"]   = df["Date"].dt.day
    df["week_of_year"]   = df["Date"].dt.isocalendar().week.astype(int)
    df["month"]          = df["Date"].dt.month
    df["quarter"]        = df["Date"].dt.quarter
    df["is_weekend"]     = (df["day_of_week"] >= 5).astype(int)
    df["is_month_start"] = df["Date"].dt.is_month_start.astype(int)
    df["is_month_end"]   = df["Date"].dt.is_month_end.astype(int)

    # Lag features (demand history)
    for lag in [1, 7, 14, 30]:
        df[f"lag_{lag}"] = df["total_demand"].shift(lag)

    # Rolling statistics (trend + volatility)
    df["rolling_mean_7d"]  = df["total_demand"].shift(1).rolling(7).mean()
    df["rolling_mean_30d"] = df["total_demand"].shift(1).rolling(30).mean()
    df["rolling_std_7d"]   = df["total_demand"].shift(1).rolling(30).std()

    # Promotion rolling effect
    df["promo_last_7d"] = (df["promo_active"]
                           .shift(1).rolling(7).sum().fillna(0))

    # Drop rows with NaN from lagging (first 30 days)
    df = df.dropna().reset_index(drop=True)
    return df

daily_fe = engineer_features(daily)
print(f"After feature engineering: {len(daily_fe)} rows "
      f"(first 30 days dropped for lag warmup)")
print(f"\nFeature columns ({len(daily_fe.columns)}):")
print(daily_fe.columns.tolist())

# ── 4. TRAIN / TEST SPLIT  (80 / 20 time-based) ──────────────
split_idx   = int(len(daily_fe) * 0.80)
train       = daily_fe.iloc[:split_idx]
test        = daily_fe.iloc[split_idx:]

print(f"\nTrain: {train['Date'].min().date()} → "
      f"{train['Date'].max().date()}  ({len(train)} days)")
print(f"Test : {test['Date'].min().date()}  → "
      f"{test['Date'].max().date()}  ({len(test)} days)")

# ── 5A. PROPHET MODEL ─────────────────────────────────────────
print("\n── Training Prophet …")

prophet_df = daily[["Date", "total_demand"]].rename(
    columns={"Date": "ds", "total_demand": "y"})

# Add regressors for external signals
prophet_df = prophet_df.merge(
    daily[["Date", "promo_active", "recall_active"]],
    left_on="ds", right_on="Date", how="left"
).drop(columns="Date")

m = Prophet(
    yearly_seasonality  = True,
    weekly_seasonality  = True,
    daily_seasonality   = False,
    seasonality_mode    = "multiplicative",  # demand spikes scale with level
    changepoint_prior_scale = 0.05,          # flexibility of trend changes
)
m.add_regressor("promo_active")
m.add_regressor("recall_active")

# Train on the same period as XGBoost for fair comparison
prophet_train = prophet_df.iloc[:split_idx + 30]  # +30 to account for lag drop
m.fit(prophet_train)

# Forecast over the test window
future   = m.make_future_dataframe(periods=len(test), freq="D")
future   = future.merge(prophet_df[["ds","promo_active","recall_active"]],
                        on="ds", how="left").fillna(0)
forecast = m.predict(future)

prophet_pred = (forecast
                .set_index("ds")["yhat"]
                .loc[test["Date"].values]
                .values
                .clip(min=0))

# ── 5B. XGBOOST MODEL ─────────────────────────────────────────
print("── Training XGBoost …")

feature_cols = [
    "day_of_week","day_of_month","week_of_year","month","quarter",
    "is_weekend","is_month_start","is_month_end",
    "lag_1","lag_7","lag_14","lag_30",
    "rolling_mean_7d","rolling_mean_30d","rolling_std_7d",
    "promo_active","recall_active","new_vehicle",
    "promo_last_7d","stockout_count"
]

X_train = train[feature_cols]
y_train = train["total_demand"]
X_test  = test[feature_cols]
y_test  = test["total_demand"]

xgb = XGBRegressor(
    n_estimators      = 500,
    learning_rate     = 0.05,
    max_depth         = 6,
    subsample         = 0.8,
    colsample_bytree  = 0.8,
    random_state      = 42,
    early_stopping_rounds = 30,
    eval_metric       = "mae",
)
xgb.fit(X_train, y_train,
        eval_set=[(X_test, y_test)],
        verbose=False)

xgb_pred = xgb.predict(X_test).clip(min=0)

# ── 6. EVALUATE BOTH MODELS ───────────────────────────────────
def evaluate(name, y_true, y_pred):
    mae   = mean_absolute_error(y_true, y_pred)
    rmse  = np.sqrt(mean_squared_error(y_true, y_pred))
    r2    = r2_score(y_true, y_pred)
    mape  = (np.abs((y_true - y_pred) / np.maximum(y_true, 1))
             .mean() * 100)
    print(f"  {name:<12}  MAE={mae:8.1f}  RMSE={rmse:8.1f}  "
          f"R²={r2:.3f}  MAPE={mape:.1f}%")
    return {"MAE": mae, "RMSE": rmse, "R2": r2, "MAPE": mape}

print("\n── Model Evaluation (Test Set) ──────────────────────────")
p_scores   = evaluate("Prophet",  y_test.values, prophet_pred)
xgb_scores = evaluate("XGBoost",  y_test.values, xgb_pred)

# ── 7. FEATURE IMPORTANCE ─────────────────────────────────────
feat_imp = (pd.Series(xgb.feature_importances_, index=feature_cols)
            .sort_values(ascending=False))

# ── 8. FORECAST NEXT 30 DAYS ─────────────────────────────────
print("\n── Generating 30-day forward forecast …")

# Build future feature rows by rolling the last known values
last_known  = daily_fe.iloc[-1].copy()
last_demand = daily_fe["total_demand"].values

future_rows = []
for i in range(1, 31):
    future_date = daily_fe["Date"].max() + pd.Timedelta(days=i)
    row = {
        "Date"          : future_date,
        "day_of_week"   : future_date.dayofweek,
        "day_of_month"  : future_date.day,
        "week_of_year"  : future_date.isocalendar()[1],
        "month"         : future_date.month,
        "quarter"       : (future_date.month - 1) // 3 + 1,
        "is_weekend"    : int(future_date.dayofweek >= 5),
        "is_month_start": int(future_date.day == 1),
        "is_month_end"  : int(future_date.day ==
                              pd.Timestamp(future_date.year,
                                           future_date.month, 1)
                              .days_in_month),
        "lag_1"         : last_demand[-1],
        "lag_7"         : last_demand[-7] if len(last_demand) >= 7  else np.mean(last_demand),
        "lag_14"        : last_demand[-14] if len(last_demand) >= 14 else np.mean(last_demand),
        "lag_30"        : last_demand[-30] if len(last_demand) >= 30 else np.mean(last_demand),
        "rolling_mean_7d" : np.mean(last_demand[-7:]),
        "rolling_mean_30d": np.mean(last_demand[-30:]),
        "rolling_std_7d"  : np.std(last_demand[-7:]),
        "promo_active"  : 0,
        "recall_active" : 0,
        "new_vehicle"   : 0,
        "promo_last_7d" : 0,
        "stockout_count": 0,
    }
    future_rows.append(row)
    # Append predicted value as next lag
    pred_val = xgb.predict(
        pd.DataFrame([row])[feature_cols])[0]
    last_demand = np.append(last_demand, pred_val)

future_df   = pd.DataFrame(future_rows)
future_preds = xgb.predict(future_df[feature_cols]).clip(min=0)
future_df["forecasted_demand"] = future_preds

print("\n30-Day Demand Forecast:")
print(future_df[["Date","forecasted_demand"]]
      .to_string(index=False))

# ── 9. PLOTS ──────────────────────────────────────────────────
fig = plt.figure(figsize=(18, 14))
fig.suptitle("MODULE 1 — Demand Forecasting",
             fontsize=16, fontweight="bold", y=1.01)
gs  = gridspec.GridSpec(2, 2, figure=fig,
                         hspace=0.45, wspace=0.35)

# Plot 1 — Actual vs Predicted (both models)
ax0 = fig.add_subplot(gs[0, :])
ax0.plot(test["Date"], y_test.values,
         label="Actual",  color="#2C3E50", linewidth=1.5)
ax0.plot(test["Date"], xgb_pred,
         label=f"XGBoost (MAPE={xgb_scores['MAPE']:.1f}%)",
         color="#E74C3C", linewidth=1.5, linestyle="--")
ax0.plot(test["Date"], prophet_pred,
         label=f"Prophet (MAPE={p_scores['MAPE']:.1f}%)",
         color="#3498DB", linewidth=1.5, linestyle=":")
ax0.fill_between(test["Date"],
                 xgb_pred * 0.92, xgb_pred * 1.08,
                 alpha=0.15, color="#E74C3C",
                 label="XGBoost ±8% band")
ax0.set_title("Actual vs Predicted Daily Demand (Test Set)",
              fontsize=12, fontweight="bold")
ax0.set_xlabel("Date"); ax0.set_ylabel("Total Demand Qty")
ax0.legend(loc="upper left"); ax0.grid(alpha=0.3)

# Plot 2 — 30-Day Forward Forecast
ax1 = fig.add_subplot(gs[1, 0])
ax1.plot(daily_fe["Date"].iloc[-60:],
         daily_fe["total_demand"].iloc[-60:],
         color="#2C3E50", label="Historical (last 60d)", linewidth=1.5)
ax1.plot(future_df["Date"], future_df["forecasted_demand"],
         color="#E74C3C", label="30-Day Forecast",
         linewidth=2, linestyle="--", marker="o", markersize=3)
ax1.fill_between(future_df["Date"],
                 future_df["forecasted_demand"] * 0.90,
                 future_df["forecasted_demand"] * 1.10,
                 alpha=0.2, color="#E74C3C", label="±10% CI")
ax1.axvline(daily_fe["Date"].max(), color="gray",
            linestyle=":", linewidth=1.5, label="Forecast start")
ax1.set_title("30-Day Forward Demand Forecast",
              fontsize=12, fontweight="bold")
ax1.set_xlabel("Date"); ax1.set_ylabel("Demand Qty")
ax1.legend(fontsize=8); ax1.grid(alpha=0.3)
plt.setp(ax1.xaxis.get_majorticklabels(), rotation=30)

# Plot 3 — XGBoost Feature Importance
ax2 = fig.add_subplot(gs[1, 1])
feat_imp.head(12).sort_values().plot(
    kind="barh", ax=ax2, color="#3498DB", edgecolor="white")
ax2.set_title("XGBoost — Top 12 Feature Importances",
              fontsize=12, fontweight="bold")
ax2.set_xlabel("Importance Score")
ax2.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.savefig("module1_forecasting.png", dpi=150, bbox_inches="tight")
plt.show()
print("\nSaved → module1_forecasting.png")

# ── 10. CATEGORY-LEVEL FORECAST SUMMARY ──────────────────────
print("\n── Category-Level Demand Summary (last 30 days) ────────")
cat_summary = (demand_clean
               .groupby("Category_Group")
               .agg(
                   Total_Demand  = ("Demand_Qty",   "sum"),
                   Avg_Daily     = ("Demand_Qty",   "mean"),
                   Total_Revenue = ("Sales_Value",  "sum"),
                   Unique_Parts  = ("Part_Number",  "nunique"),
               )
               .sort_values("Total_Revenue", ascending=False))
print(cat_summary.to_string())
print("\n✅ Module 1 complete — run Module 2 next.")
```

```python
# =============================================================
#  MODULE 2 — INVENTORY OPTIMISATION
#  EOQ + Safety Stock + Reorder Point + Stockout Risk Scoring
# =============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

# ── 1. LOAD DATA ──────────────────────────────────────────────
demand_clean = pd.read_csv("demand_clean.csv",
                           parse_dates=["Date"])

# ── 2. BUSINESS PARAMETERS ───────────────────────────────────
# Adjust these to match your actual operations
ORDER_COST       = 1500    # ₹ per order placed (admin + shipping)
HOLDING_RATE     = 0.25    # 25% of unit price per year (storage + capital)
SERVICE_LEVEL    = 0.95    # 95% → Z-score = 1.65
Z_SCORE          = 1.65
LEAD_TIME_DAYS   = {       # Days from order to shelf by category
    "Macro-Components"                  : 7,
    "Mid-Level Components"              : 5,
    "High-Velocity Workshop Consumables": 3,
    "Periodic Maintenance Kits"         : 4,
    "Micro-Components"                  : 3,
}
DEFAULT_LEAD_TIME = 5

# ── 3. PART-LEVEL AGGREGATION ─────────────────────────────────
print("── Computing part-level demand statistics …")

part_stats = (demand_clean
    .groupby(["Part_Number", "Description",
              "Category_Group", "Unit_Price"])
    .agg(
        total_demand  = ("Demand_Qty",   "sum"),
        avg_daily_dem = ("Demand_Qty",   "mean"),
        std_daily_dem = ("Demand_Qty",   "std"),
        avg_on_hand   = ("On_Hand_Qty",  "mean"),
        min_on_hand   = ("On_Hand_Qty",  "min"),
        current_rop   = ("Reorder_Point","mean"),
        current_soq   = ("SOQ_Suggested_Qty", "mean"),
        stockout_days = ("Stockout_Flag","sum"),
        obs_days       = ("Demand_Qty",   "count"),
    )
    .reset_index()
)

# Fill std=0 for parts with very consistent demand
part_stats["std_daily_dem"] = part_stats["std_daily_dem"].fillna(0)

# ── 4. EOQ CALCULATION ────────────────────────────────────────
print("── Computing EOQ …")

def compute_eoq(row):
    D = row["total_demand"]            # annual demand (units)
    S = ORDER_COST                     # order cost (₹)
    H = row["Unit_Price"] * HOLDING_RATE  # holding cost (₹/unit/year)
    if H <= 0 or D <= 0:
        return 0
    return round(np.sqrt((2 * D * S) / H), 0)

part_stats["EOQ"] = part_stats.apply(compute_eoq, axis=1)

# ── 5. SAFETY STOCK & REORDER POINT ──────────────────────────
print("── Computing Safety Stock and Reorder Points …")

def compute_safety_stock(row):
    lead_time = LEAD_TIME_DAYS.get(row["Category_Group"],
                                    DEFAULT_LEAD_TIME)
    ss = Z_SCORE * row["std_daily_dem"] * np.sqrt(lead_time)
    return round(ss, 0)

def compute_rop(row):
    lead_time = LEAD_TIME_DAYS.get(row["Category_Group"],
                                    DEFAULT_LEAD_TIME)
    ss  = compute_safety_stock(row)
    rop = (row["avg_daily_dem"] * lead_time) + ss
    return round(rop, 0)

part_stats["Safety_Stock"]  = part_stats.apply(
    compute_safety_stock, axis=1)
part_stats["Suggested_ROP"] = part_stats.apply(compute_rop, axis=1)

# ── 6. DAYS OF STOCK REMAINING ────────────────────────────────
part_stats["Days_Of_Stock"] = np.where(
    part_stats["avg_daily_dem"] > 0,
    (part_stats["avg_on_hand"] / part_stats["avg_daily_dem"]).round(1),
    999  # no demand = infinite stock
)

# ── 7. ROP DELTA — how far off is the current reorder point? ─
part_stats["ROP_Delta"] = (part_stats["Suggested_ROP"]
                           - part_stats["current_rop"]).round(0)
part_stats["ROP_Status"] = pd.cut(
    part_stats["ROP_Delta"],
    bins=[-np.inf, -10, -1, 1, 10, np.inf],
    labels=["Significantly Over-stocked ROP",
            "Slightly Over-stocked ROP",
            "Correctly Set",
            "Slightly Under-stocked ROP",
            "Significantly Under-stocked ROP"]
)

# ── 8. STOCKOUT RISK SCORE (0–100) ────────────────────────────
# Combines: stockout history, days of stock, ROP gap
def risk_score(row):
    # Component 1: historical stockout rate (0-40 pts)
    stockout_rate = row["stockout_days"] / max(row["obs_days"], 1)
    c1 = min(stockout_rate * 40, 40)

    # Component 2: days of stock remaining (0-40 pts)
    lead = LEAD_TIME_DAYS.get(row["Category_Group"], DEFAULT_LEAD_TIME)
    if row["Days_Of_Stock"] < lead:
        c2 = 40          # already below lead time — critical
    elif row["Days_Of_Stock"] < lead * 2:
        c2 = 20
    elif row["Days_Of_Stock"] < lead * 3:
        c2 = 10
    else:
        c2 = 0

    # Component 3: ROP gap (0-20 pts)
    c3 = min(max(row["ROP_Delta"] / 10, 0), 20)

    return round(c1 + c2 + c3, 1)

part_stats["Risk_Score"] = part_stats.apply(risk_score, axis=1)
part_stats["Risk_Level"]  = pd.cut(
    part_stats["Risk_Score"],
    bins=[-1, 20, 50, 75, 101],
    labels=["🟢 Low", "🟡 Medium", "🔴 High", "🚨 Critical"]
)

# ── 9. ABC ANALYSIS (by revenue contribution) ─────────────────
part_stats["Annual_Revenue"] = (part_stats["total_demand"]
                                * part_stats["Unit_Price"])
part_stats = part_stats.sort_values(
    "Annual_Revenue", ascending=False).reset_index(drop=True)

total_rev  = part_stats["Annual_Revenue"].sum()
part_stats["Cumulative_Rev_Pct"] = (
    part_stats["Annual_Revenue"].cumsum() / total_rev * 100)

part_stats["ABC_Class"] = pd.cut(
    part_stats["Cumulative_Rev_Pct"],
    bins=[0, 70, 90, 100.1],
    labels=["A — Top 70% Revenue",
            "B — Next 20% Revenue",
            "C — Bottom 10% Revenue"]
)

# ── 10. SUMMARY PRINTS ────────────────────────────────────────
print("\n" + "="*65)
print("  INVENTORY OPTIMISATION SUMMARY")
print("="*65)

print("\n── ABC Classification:")
abc = part_stats.groupby("ABC_Class", observed=True).agg(
    Parts        = ("Part_Number",    "count"),
    Total_Rev    = ("Annual_Revenue", "sum"),
    Avg_EOQ      = ("EOQ",           "mean"),
    Avg_SS       = ("Safety_Stock",  "mean"),
).round(0)
print(abc.to_string())

print("\n── ROP Status (Current vs Suggested):")
print(part_stats["ROP_Status"].value_counts().to_string())

print("\n── Stockout Risk Distribution:")
print(part_stats["Risk_Level"].value_counts().to_string())

print("\n── 🚨 Top 15 Critical/High Risk Parts (act immediately):")
urgent = part_stats[
    part_stats["Risk_Level"].isin(["🚨 Critical","🔴 High"])
].head(15)
print(urgent[["Part_Number","Description","Category_Group",
              "Days_Of_Stock","Risk_Score","Risk_Level",
              "Suggested_ROP","current_rop","EOQ",
              "Unit_Price"]].to_string(index=False))

# ── 11. PLOTS ─────────────────────────────────────────────────
fig = plt.figure(figsize=(20, 16))
fig.suptitle("MODULE 2 — Inventory Optimisation",
             fontsize=16, fontweight="bold")
gs  = gridspec.GridSpec(3, 3, figure=fig,
                        hspace=0.55, wspace=0.4)

# Plot 1 — ABC Pareto curve
ax0 = fig.add_subplot(gs[0, :2])
ax0.bar(range(len(part_stats)),
        part_stats["Annual_Revenue"],
        color=part_stats["ABC_Class"].astype(object).map({
            "A — Top 70% Revenue"    : "#E74C3C",
            "B — Next 20% Revenue"   : "#F39C12",
            "C — Bottom 10% Revenue" : "#3498DB"}).fillna("#95A5A6"),
        width=1.0, alpha=0.8)
ax0b = ax0.twinx()
ax0b.plot(part_stats["Cumulative_Rev_Pct"].values,
          color="black", linewidth=2)
ax0b.axhline(70, color="#E74C3C", linestyle="--",
             linewidth=1, alpha=0.7, label="70% (A)")
ax0b.axhline(90, color="#F39C12", linestyle="--",
             linewidth=1, alpha=0.7, label="90% (B)")
ax0b.set_ylabel("Cumulative Revenue %")
ax0b.legend(loc="center right", fontsize=9)
ax0.set_title("ABC Pareto Analysis — Revenue Contribution by Part",
              fontweight="bold")
ax0.set_xlabel("Parts (sorted by revenue)")
ax0.set_ylabel("Annual Revenue (₹)")

# Plot 2 — Risk Level by Category
ax1 = fig.add_subplot(gs[0, 2])
risk_cat = (part_stats.groupby(
    ["Category_Group","Risk_Level"], observed=True)
    .size().unstack(fill_value=0))
risk_cat.plot(kind="bar", ax=ax1, stacked=True,
    color=["#2ECC71","#F1C40F","#E74C3C","#8E44AD"],
    edgecolor="white")
ax1.set_title("Risk Distribution\nby Category",
              fontweight="bold")
ax1.set_xlabel("")
ax1.set_ylabel("Number of Parts")
ax1.tick_params(axis="x", rotation=45)
ax1.legend(title="Risk", fontsize=7,
           loc="upper right")

# Plot 3 — EOQ vs Unit Price scatter
ax2 = fig.add_subplot(gs[1, :2])
cat_colors = {
    "High-Velocity Workshop Consumables": "#3498DB",
    "Macro-Components"                  : "#E74C3C",
    "Mid-Level Components"              : "#F39C12",
    "Periodic Maintenance Kits"         : "#2ECC71",
    "Micro-Components"                  : "#9B59B6",
}
for cat, grp in part_stats.groupby("Category_Group"):
    ax2.scatter(grp["Unit_Price"], grp["EOQ"],
                label=cat, color=cat_colors.get(cat,"gray"),
                alpha=0.6, s=40)
ax2.set_title("EOQ vs Unit Price — Optimal Order Quantity by Part",
              fontweight="bold")
ax2.set_xlabel("Unit Price (₹)")
ax2.set_ylabel("Economic Order Quantity (units)")
ax2.legend(fontsize=8)
ax2.grid(alpha=0.3)

# Plot 4 — Days of Stock distribution
ax3 = fig.add_subplot(gs[1, 2])
dos_plot = part_stats[part_stats["Days_Of_Stock"] < 200]
ax3.hist(dos_plot["Days_Of_Stock"], bins=40,
         color="#3498DB", edgecolor="white", alpha=0.8)
ax3.axvline(7, color="#E74C3C", linestyle="--",
            linewidth=2, label="7-day danger zone")
ax3.axvline(30, color="#F39C12", linestyle="--",
            linewidth=2, label="30-day warning")
ax3.set_title("Days of Stock\nRemaining Distribution",
              fontweight="bold")
ax3.set_xlabel("Days of Stock")
ax3.set_ylabel("Number of Parts")
ax3.legend(fontsize=8)

# Plot 5 — Current ROP vs Suggested ROP
ax4 = fig.add_subplot(gs[2, :2])
sample = part_stats.head(40)
x = np.arange(len(sample))
w = 0.35
ax4.bar(x - w/2, sample["current_rop"],  width=w,
        label="Current ROP",   color="#95A5A6", alpha=0.8)
ax4.bar(x + w/2, sample["Suggested_ROP"], width=w,
        label="Suggested ROP", color="#E74C3C", alpha=0.8)
ax4.set_title("Current vs Suggested Reorder Points (Top 40 Parts by Revenue)",
              fontweight="bold")
ax4.set_xlabel("Part Index")
ax4.set_ylabel("Reorder Point (units)")
ax4.legend()
ax4.grid(axis="y", alpha=0.3)

# Plot 6 — Safety stock by category
ax5 = fig.add_subplot(gs[2, 2])
ss_cat = (part_stats.groupby("Category_Group")["Safety_Stock"]
          .sum().sort_values())
ss_cat.plot(kind="barh", ax=ax5, color="#9B59B6",
            edgecolor="white")
ax5.set_title("Total Safety Stock\nRequired by Category",
              fontweight="bold")
ax5.set_xlabel("Total Safety Stock (units)")
ax5.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.savefig("module2_inventory.png", dpi=150, bbox_inches="tight")
plt.show()
print("\nSaved → module2_inventory.png")

# ── 12. EXPORT FULL REPORT ────────────────────────────────────
cols_export = [
    "Part_Number","Description","Category_Group","ABC_Class",
    "Unit_Price","total_demand","avg_daily_dem","std_daily_dem",
    "avg_on_hand","Days_Of_Stock","Safety_Stock",
    "EOQ","current_rop","Suggested_ROP","ROP_Delta","ROP_Status",
    "stockout_days","Risk_Score","Risk_Level","Annual_Revenue"
]
part_stats[cols_export].to_csv(
    "module2_inventory_report.csv", index=False)
print("Saved → module2_inventory_report.csv")
print("\n✅ Module 2 complete — run Module 3 next.")
```

```python
# =============================================================
#  MODULE 3 — SALES & REVENUE ANALYSIS
#  Trends · Top Parts · Top Dealers · Category Performance
# =============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.ticker as mticker
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

# ── 1. LOAD ───────────────────────────────────────────────────
demand_clean = pd.read_csv("demand_clean.csv",
                           parse_dates=["Date"])

# Derived time columns
demand_clean["Month"]      = demand_clean["Date"].dt.month
demand_clean["Month_Name"] = demand_clean["Date"].dt.strftime("%b")
demand_clean["Week"]       = demand_clean["Date"].dt.isocalendar(
                             ).week.astype(int)
demand_clean["Quarter"]    = demand_clean["Date"].dt.quarter
demand_clean["Day_Name"]   = demand_clean["Date"].dt.strftime("%a")

print(f"Loaded: {len(demand_clean):,} rows")

# ── 2. MONTHLY REVENUE TREND ──────────────────────────────────
monthly = (demand_clean
    .groupby(["Month","Month_Name"], sort=True)
    .agg(
        Revenue      = ("Sales_Value",  "sum"),
        Demand       = ("Demand_Qty",   "sum"),
        Orders       = ("Demand_ID",    "count"),
        Unique_Parts = ("Part_Number",  "nunique"),
    )
    .reset_index()
    .sort_values("Month"))

monthly["MoM_Growth"] = monthly["Revenue"].pct_change() * 100
monthly["Avg_Order_Value"] = monthly["Revenue"] / monthly["Orders"]

print("\n── Monthly Revenue Trend ────────────────────────────────")
print(monthly[["Month_Name","Revenue","Demand",
               "Orders","MoM_Growth","Avg_Order_Value"]]
      .to_string(index=False))

# ── 3. QUARTERLY SUMMARY ──────────────────────────────────────
quarterly = (demand_clean
    .groupby("Quarter")
    .agg(Revenue=("Sales_Value","sum"),
         Demand =("Demand_Qty", "sum"),
         Orders =("Demand_ID",  "count"))
    .reset_index())

quarterly["Rev_Share_Pct"] = (quarterly["Revenue"]
                               / quarterly["Revenue"].sum() * 100).round(1)
print("\n── Quarterly Summary ────────────────────────────────────")
print(quarterly.to_string(index=False))

# ── 4. TOP PARTS BY REVENUE ───────────────────────────────────
top_parts = (demand_clean
    .groupby(["Part_Number","Description","Category_Group"])
    .agg(
        Revenue       = ("Sales_Value", "sum"),
        Units_Sold    = ("Demand_Qty",  "sum"),
        Avg_Unit_Price= ("Unit_Price",  "mean"),
        Num_Dealers   = ("Dealer_ID",   "nunique"),
    )
    .reset_index()
    .sort_values("Revenue", ascending=False))

top_parts["Revenue_Share_Pct"] = (
    top_parts["Revenue"] / top_parts["Revenue"].sum() * 100).round(2)
top_parts["Cumulative_Pct"] = top_parts["Revenue_Share_Pct"].cumsum()

print("\n── Top 20 Parts by Revenue ──────────────────────────────")
print(top_parts.head(20)[["Part_Number","Description",
                           "Category_Group","Revenue",
                           "Units_Sold","Revenue_Share_Pct",
                           "Cumulative_Pct"]].to_string(index=False))

# ── 5. TOP DEALERS BY REVENUE ─────────────────────────────────
dealer_perf = (demand_clean
    .groupby(["Dealer_ID","Dealer_Name","Zone"])
    .agg(
        Revenue      = ("Sales_Value",  "sum"),
        Demand       = ("Demand_Qty",   "sum"),
        Unique_Parts = ("Part_Number",  "nunique"),
        Orders       = ("Demand_ID",    "count"),
        Stockouts    = ("Stockout_Flag","sum"),
    )
    .reset_index()
    .sort_values("Revenue", ascending=False))

dealer_perf["Revenue_Share"]   = (
    dealer_perf["Revenue"] / dealer_perf["Revenue"].sum() * 100).round(2)
dealer_perf["Stockout_Rate"]   = (
    dealer_perf["Stockouts"] / dealer_perf["Orders"] * 100).round(1)
dealer_perf["Avg_Order_Value"] = (
    dealer_perf["Revenue"] / dealer_perf["Orders"]).round(0)

print("\n── Top 20 Dealers by Revenue ────────────────────────────")
print(dealer_perf.head(20)[["Dealer_Name","Zone","Revenue",
                              "Revenue_Share","Unique_Parts",
                              "Stockout_Rate",
                              "Avg_Order_Value"]].to_string(index=False))

# ── 6. ZONE PERFORMANCE ───────────────────────────────────────
zone_perf = (demand_clean
    .groupby("Zone")
    .agg(Revenue   =("Sales_Value","sum"),
         Demand    =("Demand_Qty", "sum"),
         Dealers   =("Dealer_ID",  "nunique"),
         Stockouts =("Stockout_Flag","sum"))
    .reset_index()
    .sort_values("Revenue", ascending=False))
zone_perf["Rev_Per_Dealer"] = (
    zone_perf["Revenue"] / zone_perf["Dealers"]).round(0)
zone_perf["Stockout_Rate"]  = (
    zone_perf["Stockouts"] /
    demand_clean.groupby("Zone")["Demand_ID"].count().values * 100
).round(1)

print("\n── Zone Performance ─────────────────────────────────────")
print(zone_perf.to_string(index=False))

# ── 7. CATEGORY MONTHLY HEATMAP DATA ─────────────────────────
cat_monthly = (demand_clean
    .groupby(["Category_Group","Month_Name","Month"])["Sales_Value"]
    .sum()
    .reset_index()
    .sort_values("Month"))

cat_pivot = cat_monthly.pivot(index="Category_Group",
                               columns="Month_Name",
                               values="Sales_Value")
month_order = ["Jan","Feb","Mar","Apr","May","Jun",
               "Jul","Aug","Sep","Oct","Nov","Dec"]
cat_pivot   = cat_pivot[
    [m for m in month_order if m in cat_pivot.columns]]

# ── 8. DAY-OF-WEEK DEMAND PATTERN ────────────────────────────
dow_pattern = (demand_clean
    .groupby("Day_Name")
    .agg(Avg_Revenue=("Sales_Value","mean"),
         Avg_Demand =("Demand_Qty", "mean"))
    .reindex(["Mon","Tue","Wed","Thu","Fri","Sat","Sun"])
    .reset_index())

# ── 9. PROMO IMPACT ANALYSIS ─────────────────────────────────
promo_impact = (demand_clean
    .groupby("Promotion_Flag")
    .agg(
        Avg_Demand  =("Demand_Qty", "mean"),
        Avg_Revenue =("Sales_Value","mean"),
        Count       =("Demand_ID",  "count"),
    )
    .reset_index())
promo_impact["Promotion_Flag"] = promo_impact["Promotion_Flag"].map(
    {0:"No Promo", 1:"Promo Active"})
print("\n── Promotion Impact ─────────────────────────────────────")
print(promo_impact.to_string(index=False))

# ── 10. PLOTS ─────────────────────────────────────────────────
fig = plt.figure(figsize=(22, 20))
fig.suptitle("MODULE 3 — Sales & Revenue Analysis",
             fontsize=17, fontweight="bold")
gs  = gridspec.GridSpec(3, 3, figure=fig,
                        hspace=0.55, wspace=0.4)

# Plot 1 — Monthly Revenue + MoM growth
ax0 = fig.add_subplot(gs[0, :2])
bars = ax0.bar(monthly["Month_Name"], monthly["Revenue"] / 1e6,
               color="#3498DB", alpha=0.8, edgecolor="white",
               label="Monthly Revenue")
ax0b = ax0.twinx()
ax0b.plot(monthly["Month_Name"], monthly["MoM_Growth"],
          color="#E74C3C", marker="o", linewidth=2,
          label="MoM Growth %")
ax0b.axhline(0, color="gray", linestyle="--", linewidth=1)
ax0b.set_ylabel("Month-on-Month Growth (%)", color="#E74C3C")
ax0.set_title("Monthly Revenue Trend with MoM Growth",
              fontweight="bold", fontsize=12)
ax0.set_xlabel("Month")
ax0.set_ylabel("Revenue (₹ Millions)")
ax0.yaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"₹{x:.0f}M"))
lines1, labels1 = ax0.get_legend_handles_labels()
lines2, labels2 = ax0b.get_legend_handles_labels()
ax0.legend(lines1 + lines2, labels1 + labels2, loc="upper left")
ax0.grid(axis="y", alpha=0.3)

# Plot 2 — Quarterly donut
ax1 = fig.add_subplot(gs[0, 2])
colors_q = ["#2ECC71","#3498DB","#E74C3C","#F39C12"]
wedges, texts, autotexts = ax1.pie(
    quarterly["Revenue"],
    labels=[f"Q{q}" for q in quarterly["Quarter"]],
    autopct="%1.1f%%",
    colors=colors_q,
    startangle=90,
    wedgeprops=dict(width=0.5))
ax1.set_title("Revenue by Quarter", fontweight="bold", fontsize=12)

# Plot 3 — Top 15 Parts by Revenue
ax2 = fig.add_subplot(gs[1, :2])
top15 = top_parts.head(15).sort_values("Revenue")
colors_cat = {
    "High-Velocity Workshop Consumables": "#3498DB",
    "Macro-Components"                  : "#E74C3C",
    "Mid-Level Components"              : "#F39C12",
    "Periodic Maintenance Kits"         : "#2ECC71",
    "Micro-Components"                  : "#9B59B6",
}
bar_colors = top15["Category_Group"].map(colors_cat).fillna("gray")
bars2 = ax2.barh(top15["Description"], top15["Revenue"] / 1e6,
                 color=bar_colors, edgecolor="white", alpha=0.85)
ax2.set_title("Top 15 Parts by Revenue",
              fontweight="bold", fontsize=12)
ax2.set_xlabel("Revenue (₹ Millions)")
ax2.xaxis.set_major_formatter(
    mticker.FuncFormatter(lambda x, _: f"₹{x:.0f}M"))
for bar, (_, row) in zip(bars2, top15.iterrows()):
    ax2.text(bar.get_width() + 0.1,
             bar.get_y() + bar.get_height() / 2,
             f"{row['Revenue_Share_Pct']:.1f}%",
             va="center", fontsize=8)
ax2.grid(axis="x", alpha=0.3)

# Plot 4 — Top 10 Dealers
ax3 = fig.add_subplot(gs[1, 2])
top10d = dealer_perf.head(10).sort_values("Revenue")
ax3.barh(top10d["Dealer_Name"],
         top10d["Revenue"] / 1e6,
         color="#9B59B6", edgecolor="white", alpha=0.85)
ax3.set_title("Top 10 Dealers\nby Revenue",
              fontweight="bold", fontsize=12)
ax3.set_xlabel("Revenue (₹ Millions)")
ax3.grid(axis="x", alpha=0.3)

# Plot 5 — Category × Month heatmap
ax4 = fig.add_subplot(gs[2, :2])
sns.heatmap(cat_pivot / 1e6, annot=True, fmt=".1f",
            cmap="YlOrRd", linewidths=0.5,
            linecolor="white", ax=ax4,
            cbar_kws={"label": "Revenue (₹M)"})
ax4.set_title("Revenue Heatmap — Category × Month (₹ Millions)",
              fontweight="bold", fontsize=12)
ax4.set_xlabel("Month")
ax4.set_ylabel("")

# Plot 6 — Day-of-Week pattern + Zone revenue
ax5 = fig.add_subplot(gs[2, 2])
ax5.bar(dow_pattern["Day_Name"],
        dow_pattern["Avg_Revenue"],
        color=["#E74C3C" if d in ["Sat","Sun"] else "#3498DB"
               for d in dow_pattern["Day_Name"]],
        edgecolor="white", alpha=0.85)
ax5.set_title("Avg Daily Revenue\nby Day of Week",
              fontweight="bold", fontsize=12)
ax5.set_xlabel("Day")
ax5.set_ylabel("Avg Revenue (₹)")
ax5.grid(axis="y", alpha=0.3)

plt.tight_layout()
plt.savefig("module3_sales.png", dpi=150, bbox_inches="tight")
plt.show()
print("\nSaved → module3_sales.png")

# ── 11. EXPORT ────────────────────────────────────────────────
top_parts.to_csv("module3_top_parts.csv",   index=False)
dealer_perf.to_csv("module3_dealer_perf.csv", index=False)
monthly.to_csv("module3_monthly_trend.csv",  index=False)
print("Saved → module3_top_parts.csv")
print("Saved → module3_dealer_perf.csv")
print("Saved → module3_monthly_trend.csv")
print("\n✅ Module 3 complete — run Module 4 next.")
```

```python
# =============================================================
#  MODULE 4 — SERVICE PATTERN ANALYSIS
#  Job Cards · Vehicle Models · Customer Behaviour ·
#  Part Consumption Patterns
# =============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import matplotlib.ticker as mticker
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

# ── 1. LOAD ALL FILES ─────────────────────────────────────────
demand_clean = pd.read_csv("demand_clean.csv",
                           parse_dates=["Date"])
jc_hdr       = pd.read_csv("job_card_header.csv",
                           parse_dates=["Date_In"])
jc_lines     = pd.read_csv("/content/job_card_line_items.csv")
vehicle_df   = pd.read_csv("/content/vehicle_master.csv")
customer_df  = pd.read_csv("/content/customer_master.csv")

print(f"Job Card Headers    : {len(jc_hdr):,} rows")
print(f"Job Card Line Items : {len(jc_lines):,} rows")
print(f"Vehicle Master      : {len(vehicle_df):,} rows")
print(f"Customer Master     : {len(customer_df):,} rows")

# Time features on job cards
jc_hdr["Month"]      = jc_hdr["Date_In"].dt.month
jc_hdr["Month_Name"] = jc_hdr["Date_In"].dt.strftime("%b")
jc_hdr["Quarter"]    = jc_hdr["Date_In"].dt.quarter
jc_hdr["Day_Name"]   = jc_hdr["Date_In"].dt.strftime("%a")
jc_hdr["Week"]       = jc_hdr["Date_In"].dt.isocalendar(
                        ).week.astype(int)

# ── 2. MERGE JOB CARDS WITH LINE ITEMS ───────────────────────
print("\n── Merging job cards with line items …")
jc_full = jc_lines.merge(
    jc_hdr[["Job_Card_Number","Date_In","Dealer_ID",
             "Chassis_Number","Customer_ID",
             "Month","Month_Name","Quarter",
             "Day_Name","Week"]],
    on="Job_Card_Number", how="inner"
)
print(f"Merged dataset      : {len(jc_full):,} rows")

# ── 3. MERGE WITH VEHICLE MASTER ─────────────────────────────
print("── Merging with vehicle master …")
print(f"Vehicle master cols : {vehicle_df.columns.tolist()}")

# Flexible join — find the chassis column automatically
chassis_col_v = [c for c in vehicle_df.columns
                 if "chassis" in c.lower() or "vin" in c.lower()]
if chassis_col_v:
    jc_full = jc_full.merge(
        vehicle_df.rename(columns={chassis_col_v[0]: "Chassis_Number"}),
        on="Chassis_Number", how="left"
    )
    print(f"Vehicle data joined on: {chassis_col_v[0]}")
else:
    print("  ⚠️  No chassis column found in vehicle_master — skipping join")

# ── 4. JOB CARD VOLUME ANALYSIS ───────────────────────────────
print("\n" + "="*65)
print("  SERVICE VOLUME ANALYSIS")
print("="*65)

# Monthly job card volume
monthly_jc = (jc_hdr
    .groupby(["Month","Month_Name"])
    .agg(
        Job_Cards        = ("Job_Card_Number", "count"),
        Unique_Customers = ("Customer_ID",     "nunique"),
        Unique_Vehicles  = ("Chassis_Number",  "nunique"),
        Unique_Dealers   = ("Dealer_ID",       "nunique"),
    )
    .reset_index()
    .sort_values("Month"))

monthly_jc["MoM_Growth"] = (monthly_jc["Job_Cards"]
                             .pct_change() * 100).round(1)
monthly_jc["Avg_Cards_Per_Dealer"] = (
    monthly_jc["Job_Cards"] /
    monthly_jc["Unique_Dealers"]).round(1)

print("\n── Monthly Service Volume:")
print(monthly_jc[["Month_Name","Job_Cards","Unique_Customers",
                   "Unique_Vehicles","MoM_Growth",
                   "Avg_Cards_Per_Dealer"]].to_string(index=False))

# Day of week pattern
dow_jc = (jc_hdr
    .groupby("Day_Name")["Job_Card_Number"]
    .count()
    .reindex(["Mon","Tue","Wed","Thu","Fri","Sat","Sun"])
    .reset_index()
    .rename(columns={"Job_Card_Number":"Job_Cards"}))

print("\n── Job Cards by Day of Week:")
print(dow_jc.to_string(index=False))

# ── 5. PARTS CONSUMPTION ANALYSIS ────────────────────────────
print("\n" + "="*65)
print("  PARTS CONSUMPTION FROM JOB CARDS")
print("="*65)

part_consumption = (jc_full
    .groupby(["Part_Number","Description"])
    .agg(
        Times_Used      = ("Line_Item_ID",       "count"),
        Total_Qty       = ("Quantity_Consumed",   "sum"),
        Total_Billed    = ("Billed_Amount",       "sum"),
        Avg_Qty_Per_Job = ("Quantity_Consumed",   "mean"),
        Unique_Dealers  = ("Dealer_ID",           "nunique"),
    )
    .reset_index()
    .sort_values("Total_Billed", ascending=False))

print("\n── Top 20 Parts by Billed Amount (from Job Cards):")
print(part_consumption.head(20)[["Part_Number","Description",
                                  "Times_Used","Total_Qty",
                                  "Total_Billed","Avg_Qty_Per_Job",
                                  "Unique_Dealers"]]
      .to_string(index=False))

# ── 6. VEHICLE MODEL SERVICE PATTERNS ────────────────────────
print("\n" + "="*65)
print("  VEHICLE MODEL SERVICE PATTERNS")
print("="*65)

# Check which vehicle columns are available after merge
vehicle_cols = [c for c in jc_full.columns
                if any(k in c.lower() for k in
                       ["model","variant","fuel","vehicle"])]
print(f"Available vehicle columns: {vehicle_cols}")

if vehicle_cols:
    model_col = vehicle_cols[0]   # use first available model column
    model_service = (jc_full
        .groupby(model_col)
        .agg(
            Service_Visits  = ("Job_Card_Number", "nunique"),
            Total_Parts_Used= ("Quantity_Consumed","sum"),
            Total_Revenue   = ("Billed_Amount",    "sum"),
            Unique_Parts    = ("Part_Number",      "nunique"),
            Avg_Parts_Per_Visit = ("Quantity_Consumed","mean"),
        )
        .reset_index()
        .sort_values("Service_Visits", ascending=False))

    print(f"\n── Service Frequency by {model_col} (Top 15):")
    print(model_service.head(15).to_string(index=False))

# ── 7. CUSTOMER BEHAVIOUR ANALYSIS ───────────────────────────
print("\n" + "="*65)
print("  CUSTOMER BEHAVIOUR ANALYSIS")
print("="*65)

customer_visits = (jc_hdr
    .groupby("Customer_ID")
    .agg(
        Total_Visits    = ("Job_Card_Number", "count"),
        First_Visit     = ("Date_In",         "min"),
        Last_Visit      = ("Date_In",         "max"),
        Unique_Dealers  = ("Dealer_ID",       "nunique"),
    )
    .reset_index())

customer_visits["Days_Between_Visits"] = (
    (customer_visits["Last_Visit"] -
     customer_visits["First_Visit"]).dt.days /
    customer_visits["Total_Visits"].clip(lower=1)).round(0)

# Segment customers by visit frequency
customer_visits["Loyalty_Segment"] = pd.cut(
    customer_visits["Total_Visits"],
    bins=[0, 1, 3, 6, 999],
    labels=["One-Time (1)",
            "Occasional (2-3)",
            "Regular (4-6)",
            "Loyal (7+)"])

print("\n── Customer Loyalty Segmentation:")
loyalty = (customer_visits
    .groupby("Loyalty_Segment", observed=True)
    .agg(
        Customers    = ("Customer_ID",           "count"),
        Avg_Visits   = ("Total_Visits",          "mean"),
        Avg_Gap_Days = ("Days_Between_Visits",   "mean"),
    )
    .reset_index())
loyalty["Customer_Share_Pct"] = (
    loyalty["Customers"] /
    loyalty["Customers"].sum() * 100).round(1)
print(loyalty.to_string(index=False))

# Multi-dealer customers (loyalty risk)
multi_dealer = customer_visits[
    customer_visits["Unique_Dealers"] > 1]
print(f"\n── Multi-Dealer Customers (loyalty risk):")
print(f"   Count  : {len(multi_dealer):,}")
print(f"   Pct    : {len(multi_dealer)/len(customer_visits)*100:.1f}%")
print(f"   Avg visits: {multi_dealer['Total_Visits'].mean():.1f}")

# ── 8. DEALER SERVICE EFFICIENCY ─────────────────────────────
print("\n" + "="*65)
print("  DEALER SERVICE EFFICIENCY")
print("="*65)

dealer_service = (jc_hdr
    .groupby("Dealer_ID")
    .agg(
        Total_Jobs      = ("Job_Card_Number","count"),
        Unique_Customers= ("Customer_ID",    "nunique"),
        Unique_Vehicles = ("Chassis_Number", "nunique"),
    )
    .reset_index())

# Join billed amount from line items
dealer_billing = (jc_full
    .groupby("Dealer_ID")
    .agg(Total_Billed   = ("Billed_Amount",   "sum"),
         Total_Parts    = ("Quantity_Consumed","sum"),
         Avg_Parts_Job  = ("Quantity_Consumed","mean"))
    .reset_index())

dealer_service = dealer_service.merge(dealer_billing,
                                      on="Dealer_ID", how="left")
dealer_service["Revenue_Per_Job"] = (
    dealer_service["Total_Billed"] /
    dealer_service["Total_Jobs"]).round(0)
dealer_service["Customer_Retention"] = (
    dealer_service["Unique_Customers"] /
    dealer_service["Total_Jobs"] * 100).round(1)

print("\n── Top 15 Dealers by Service Revenue:")
print(dealer_service.sort_values(
    "Total_Billed", ascending=False).head(15).to_string(index=False))

# ── 9. REPEAT VISIT INTERVAL ANALYSIS ────────────────────────
print("\n" + "="*65)
print("  SERVICE INTERVAL ANALYSIS")
print("="*65)

# Sort by customer + date to compute gaps between visits
jc_sorted = (jc_hdr[["Customer_ID","Date_In"]]
             .sort_values(["Customer_ID","Date_In"]))
jc_sorted["Prev_Visit"] = (jc_sorted
    .groupby("Customer_ID")["Date_In"].shift(1))
jc_sorted["Visit_Gap_Days"] = (
    jc_sorted["Date_In"] -
    jc_sorted["Prev_Visit"]).dt.days

gaps = jc_sorted["Visit_Gap_Days"].dropna()
print(f"\n── Service Interval Statistics (days between visits):")
print(f"   Mean   : {gaps.mean():.0f} days")
print(f"   Median : {gaps.median():.0f} days")
print(f"   Std    : {gaps.std():.0f} days")
print(f"   Min    : {gaps.min():.0f} days")
print(f"   Max    : {gaps.max():.0f} days")

interval_bins = pd.cut(gaps,
    bins=[0,30,60,90,180,365,9999],
    labels=["<30d","30-60d","60-90d",
            "90-180d","180-365d",">365d"])
print(f"\n── Visit Gap Distribution:")
print(interval_bins.value_counts().sort_index().to_string())

# ── 10. PLOTS ─────────────────────────────────────────────────
fig = plt.figure(figsize=(22, 20))
fig.suptitle("MODULE 4 — Service Pattern Analysis",
             fontsize=17, fontweight="bold")
gs  = gridspec.GridSpec(3, 3, figure=fig,
                        hspace=0.55, wspace=0.4)

# Plot 1 — Monthly job card volume + MoM growth
ax0 = fig.add_subplot(gs[0, :2])
ax0.bar(monthly_jc["Month_Name"],
        monthly_jc["Job_Cards"],
        color="#3498DB", alpha=0.85, edgecolor="white",
        label="Job Cards")
ax0b = ax0.twinx()
ax0b.plot(monthly_jc["Month_Name"],
          monthly_jc["MoM_Growth"],
          color="#E74C3C", marker="o",
          linewidth=2, label="MoM Growth %")
ax0b.axhline(0, color="gray", linestyle="--", linewidth=1)
ax0b.set_ylabel("MoM Growth (%)", color="#E74C3C")
ax0.set_title("Monthly Service Volume (Job Cards) + MoM Growth",
              fontweight="bold", fontsize=12)
ax0.set_ylabel("Number of Job Cards")
lines1, labels1 = ax0.get_legend_handles_labels()
lines2, labels2 = ax0b.get_legend_handles_labels()
ax0.legend(lines1+lines2, labels1+labels2, loc="upper left")
ax0.grid(axis="y", alpha=0.3)

# Plot 2 — Customer loyalty donut
ax1 = fig.add_subplot(gs[0, 2])
ax1.pie(loyalty["Customers"],
        labels=loyalty["Loyalty_Segment"],
        autopct="%1.1f%%",
        colors=["#E74C3C","#F39C12","#3498DB","#2ECC71"],
        startangle=90,
        wedgeprops=dict(width=0.5))
ax1.set_title("Customer Loyalty\nSegmentation",
              fontweight="bold", fontsize=12)

# Plot 3 — Top 15 parts consumed
ax2 = fig.add_subplot(gs[1, :2])
top15p = part_consumption.head(15).sort_values("Total_Billed")
ax2.barh(top15p["Description"],
         top15p["Total_Billed"] / 1e6,
         color="#E74C3C", edgecolor="white", alpha=0.85)
ax2.set_title("Top 15 Parts by Billed Amount (Job Cards)",
              fontweight="bold", fontsize=12)
ax2.set_xlabel("Total Billed (₹ Millions)")
ax2.grid(axis="x", alpha=0.3)

# Plot 4 — Day of week job card pattern
ax3 = fig.add_subplot(gs[1, 2])
ax3.bar(dow_jc["Day_Name"], dow_jc["Job_Cards"],
        color=["#E74C3C" if d in ["Sat","Sun"]
               else "#3498DB"
               for d in dow_jc["Day_Name"]],
        edgecolor="white", alpha=0.85)
ax3.set_title("Job Cards by\nDay of Week",
              fontweight="bold", fontsize=12)
ax3.set_ylabel("Number of Job Cards")
ax3.grid(axis="y", alpha=0.3)

# Plot 5 — Service interval histogram
ax4 = fig.add_subplot(gs[2, :2])
ax4.hist(gaps.clip(upper=365), bins=50,
         color="#9B59B6", edgecolor="white", alpha=0.85)
ax4.axvline(gaps.median(), color="#E74C3C",
            linewidth=2, linestyle="--",
            label=f"Median: {gaps.median():.0f}d")
ax4.axvline(90, color="#F39C12",
            linewidth=2, linestyle="--",
            label="90-day service target")
ax4.set_title("Distribution of Days Between Customer Visits",
              fontweight="bold", fontsize=12)
ax4.set_xlabel("Days Between Visits (capped at 365)")
ax4.set_ylabel("Number of Visits")
ax4.legend(); ax4.grid(alpha=0.3)

# Plot 6 — Vehicle model service frequency (if available)
ax5 = fig.add_subplot(gs[2, 2])
if vehicle_cols and "model_service" in dir():
    top_models = model_service.head(10).sort_values(
        "Service_Visits")
    ax5.barh(top_models[model_col].astype(str),
             top_models["Service_Visits"],
             color="#2ECC71", edgecolor="white", alpha=0.85)
    ax5.set_title(f"Top 10 Vehicle Models\nby Service Visits",
                  fontweight="bold", fontsize=12)
    ax5.set_xlabel("Service Visits")
    ax5.grid(axis="x", alpha=0.3)
else:
    # Fallback — dealer service efficiency
    top_dealers = dealer_service.sort_values(
        "Revenue_Per_Job", ascending=False).head(10)
    ax5.barh(top_dealers["Dealer_ID"].astype(str),
             top_dealers["Revenue_Per_Job"],
             color="#2ECC71", edgecolor="white", alpha=0.85)
    ax5.set_title("Revenue per Job Card\n(Top 10 Dealers)",
                  fontweight="bold", fontsize=12)
    ax5.set_xlabel("Avg Revenue per Job (₹)")
    ax5.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.savefig("module4_service.png", dpi=150, bbox_inches="tight")
plt.show()
print("\nSaved → module4_service.png")

# ── 11. EXPORT ────────────────────────────────────────────────
part_consumption.to_csv("module4_parts_consumed.csv",   index=False)
customer_visits.to_csv("module4_customer_segments.csv", index=False)
dealer_service.to_csv("module4_dealer_service.csv",     index=False)
monthly_jc.to_csv("module4_monthly_volume.csv",         index=False)
print("Saved → module4_parts_consumed.csv")
print("Saved → module4_customer_segments.csv")
print("Saved → module4_dealer_service.csv")
print("Saved → module4_monthly_volume.csv")
print("\n✅ Module 4 complete — all 4 modules done!")

# ── 12. MASTER SUMMARY ────────────────────────────────────────
print("\n" + "="*65)
print("  COMPLETE PROJECT SUMMARY")
print("="*65)
print("""
  MODULE 1 — Demand Forecasting
    ✅ XGBoost MAPE = 6.0%  (production-grade)
    ✅ 30-day forward forecast generated
    ✅ Key drivers: lag features + rolling means

  MODULE 2 — Inventory Optimisation
    🚨 100% of parts have under-set Reorder Points
    🚨 8 Macro-Component parts below lead-time stock
    ✅ EOQ + Safety Stock computed for all 609 parts
    ✅ Full inventory report saved to CSV

  MODULE 3 — Sales & Revenue Analysis
    🚨 South Zone: 94% stockout rate
    ⚠️  Promotions used on only 6.1% of records
    ✅ +63.6% demand lift on promo days
    ✅ Top parts, dealers, zones ranked

  MODULE 4 — Service Pattern Analysis
    ✅ Customer loyalty segments computed
    ✅ Parts consumption from job cards ranked
    ✅ Vehicle model service frequency mapped
    ✅ Service interval distribution analysed
""")
```

```python
# =============================================================
#  HEURISTIC TRANSFER UTILITY SCORE
#  "Should we move inventory from Dealer A → Dealer B?"
#
#  Scores every possible (Donor, Receiver, Part) triplet
#  and surfaces the highest-value transfers to act on.
# =============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

# ── 1. LOAD DATA ──────────────────────────────────────────────
demand_clean  = pd.read_csv("demand_clean.csv",
                            parse_dates=["Date"])
inv_report    = pd.read_csv("module2_inventory_report.csv")
dealer_perf   = pd.read_csv("module3_dealer_perf.csv")

print(f"Demand rows   : {len(demand_clean):,}")
print(f"Parts tracked : {len(inv_report):,}")
print(f"Dealers       : {len(dealer_perf):,}")

# ── 2. BUSINESS PARAMETERS ───────────────────────────────────
TRANSFER_THRESHOLD   = 55    # Score ≥ this → recommend transfer
HOLDING_RATE         = 0.25  # 25% annual holding cost
TRANSFER_COST_SAME   = 500   # ₹ cost to transfer within same zone
TRANSFER_COST_DIFF   = 1500  # ₹ cost to transfer across zones
LEAD_TIME_DAYS = {
    "Macro-Components"                  : 7,
    "Mid-Level Components"              : 5,
    "High-Velocity Workshop Consumables": 3,
    "Periodic Maintenance Kits"         : 4,
    "Micro-Components"                  : 3,
}
DEFAULT_LEAD = 5

# ── 3. BUILD DEALER × PART SNAPSHOT ──────────────────────────
# Latest stock position per dealer per part
print("\n── Building dealer × part inventory snapshot …")

# Use last 30 days for demand rate (recent signal)
recent_cutoff = demand_clean["Date"].max() - pd.Timedelta(days=30)
recent        = demand_clean[demand_clean["Date"] >= recent_cutoff]

dealer_part = (demand_clean
    .groupby(["Dealer_ID","Dealer_Name","Zone",
              "Part_Number","Description","Category_Group",
              "Unit_Price"])
    .agg(
        # Stock: use the most recent On_Hand_Qty reading
        Current_Stock    = ("On_Hand_Qty",       "last"),
        Avg_Stock        = ("On_Hand_Qty",        "mean"),
        # Demand rate over full period
        Avg_Daily_Demand = ("Demand_Qty",         "mean"),
        Std_Daily_Demand = ("Demand_Qty",         "std"),
        Total_Demand     = ("Demand_Qty",         "sum"),
        # Risk signals
        Stockout_Days    = ("Stockout_Flag",      "sum"),
        Total_Days       = ("Demand_Qty",         "count"),
        Current_ROP      = ("Reorder_Point",      "last"),
    )
    .reset_index())

dealer_part["Std_Daily_Demand"] = (dealer_part["Std_Daily_Demand"]
                                    .fillna(0))

# Days of stock remaining
dealer_part["Days_Of_Stock"] = np.where(
    dealer_part["Avg_Daily_Demand"] > 0,
    (dealer_part["Current_Stock"] /
     dealer_part["Avg_Daily_Demand"]).round(1),
    999)

# Lead time lookup
dealer_part["Lead_Time"] = dealer_part["Category_Group"].map(
    LEAD_TIME_DAYS).fillna(DEFAULT_LEAD)

# Safety stock per dealer-part
dealer_part["Safety_Stock"] = (
    1.65 * dealer_part["Std_Daily_Demand"] *
    np.sqrt(dealer_part["Lead_Time"])).round(1)

# Stockout rate
dealer_part["Stockout_Rate"] = (
    dealer_part["Stockout_Days"] /
    dealer_part["Total_Days"].clip(lower=1))

print(f"  Dealer × Part combinations : {len(dealer_part):,}")

# ── 4. COMPONENT A — RECEIVER URGENCY SCORE (0–25) ───────────
# High score = this dealer-part pair NEEDS stock urgently
def receiver_urgency(row):
    """
    Measures how desperately Dealer B needs this part.
    Sub-components:
      (a) Days of stock vs lead time    → 0–10 pts
      (b) Historical stockout rate      → 0–10 pts
      (c) Demand rate (high movers)     → 0–5  pts
    """
    lead = row["Lead_Time"]

    # (a) Proximity to stockout
    dos = row["Days_Of_Stock"]
    if dos <= 0:
        a = 10       # already stocked out
    elif dos <= lead:
        a = 8        # within lead time — critical
    elif dos <= lead * 1.5:
        a = 6        # approaching danger zone
    elif dos <= lead * 2:
        a = 4        # below safety buffer
    elif dos <= lead * 3:
        a = 2        # low but not urgent
    else:
        a = 0        # well stocked

    # (b) Historical stockout rate
    sr = row["Stockout_Rate"]
    b  = min(sr * 10, 10)    # scales linearly to max 10

    # (c) Demand velocity (high-demand parts matter more)
    avg_d = row["Avg_Daily_Demand"]
    if avg_d > 10:
        c = 5
    elif avg_d > 5:
        c = 3
    elif avg_d > 1:
        c = 1
    else:
        c = 0

    return round(a + b + c, 2)

dealer_part["Urgency_Score"] = dealer_part.apply(
    receiver_urgency, axis=1)

# ── 5. COMPONENT B — DONOR SURPLUS SCORE (0–25) ──────────────
# High score = Dealer A has excess stock it can safely give away
def donor_surplus(row):
    """
    Measures how safely Dealer A can give up units.
    Sub-components:
      (a) Surplus above ROP + Safety Stock  → 0–15 pts
      (b) Low stockout history              → 0–5  pts
      (c) Holding cost relief               → 0–5  pts
    """
    lead    = row["Lead_Time"]
    ss      = row["Safety_Stock"]
    rop     = row["Current_ROP"]
    stock   = row["Current_Stock"]
    avg_d   = row["Avg_Daily_Demand"]

    # Minimum safe stock = ROP + safety stock
    min_safe = rop + ss
    surplus  = stock - min_safe

    # (a) How much surplus relative to lead-time demand?
    lead_demand = avg_d * lead if avg_d > 0 else 1
    surplus_ratio = surplus / max(lead_demand, 1)

    if surplus_ratio >= 3:
        a = 15
    elif surplus_ratio >= 2:
        a = 12
    elif surplus_ratio >= 1:
        a = 8
    elif surplus_ratio >= 0.5:
        a = 4
    elif surplus_ratio >= 0:
        a = 1
    else:
        a = 0   # no surplus — cannot donate

    # (b) Low stockout history = safe to give away
    sr = row["Stockout_Rate"]
    b  = max(5 - sr * 10, 0)   # low stockout → high score

    # (c) Holding cost relief: high-value parts cost more to hold
    annual_holding = row["Unit_Price"] * HOLDING_RATE
    if annual_holding > 30000:
        c = 5
    elif annual_holding > 10000:
        c = 3
    elif annual_holding > 2000:
        c = 1
    else:
        c = 0

    return round(max(a + b + c, 0), 2), round(max(surplus, 0), 0)

dealer_part[["Surplus_Score","Transferable_Units"]] = (
    dealer_part.apply(
        lambda r: pd.Series(donor_surplus(r)), axis=1))

# ── 6. COMPONENT C — PART CRITICALITY SCORE (0–25) ───────────
# High score = this part is important enough to prioritise
def part_criticality(row):
    """
    How important is this part to the business?
    Sub-components:
      (a) ABC class from revenue          → 0–15 pts
      (b) Category (Macro = highest risk) → 0–5  pts
      (c) Breadth of demand (all dealers) → 0–5  pts
    """
    # (a) Revenue importance — use unit price as proxy
    price = row["Unit_Price"]
    if price > 100000:
        a = 15     # Macro-Component
    elif price > 30000:
        a = 12
    elif price > 10000:
        a = 8
    elif price > 2000:
        a = 5
    else:
        a = 2

    # (b) Category risk
    cat = row["Category_Group"]
    cat_map = {
        "Macro-Components"                  : 5,
        "Mid-Level Components"              : 4,
        "Periodic Maintenance Kits"         : 3,
        "High-Velocity Workshop Consumables": 2,
        "Micro-Components"                  : 1,
    }
    b = cat_map.get(cat, 2)

    # (c) Demand breadth — high avg demand = widely needed
    avg_d = row["Avg_Daily_Demand"]
    if avg_d > 10:
        c = 5
    elif avg_d > 5:
        c = 3
    elif avg_d > 1:
        c = 2
    else:
        c = 1

    return round(a + b + c, 2)

dealer_part["Criticality_Score"] = dealer_part.apply(
    part_criticality, axis=1)

# ── 7. COMPONENT D — TRANSFER FRICTION SCORE (0–25) ──────────
# High score = HIGH friction = penalises the transfer
# (this is SUBTRACTED from total)
def transfer_friction(donor_zone, receiver_zone,
                      unit_price, transfer_units):
    """
    How costly/risky is this transfer?
    Sub-components:
      (a) Geographic distance (zone match) → 0–10 pts
      (b) Transfer cost vs part value      → 0–10 pts
      (c) Minimum viable qty               → 0–5  pts
    """
    # (a) Zone friction
    a = 0 if donor_zone == receiver_zone else 8

    # (b) Transfer cost as % of part value
    cost   = (TRANSFER_COST_SAME
              if donor_zone == receiver_zone
              else TRANSFER_COST_DIFF)
    pct    = cost / max(unit_price, 1) * 100
    if pct > 5:
        b = 10     # cost > 5% of part value — not worth it
    elif pct > 2:
        b = 6
    elif pct > 0.5:
        b = 3
    else:
        b = 0      # negligible cost vs part value

    # (c) Minimum viable quantity (moving 0.5 units = impractical)
    c = 0 if transfer_units >= 1 else 5

    return round(a + b + c, 2)

# ── 8. GENERATE ALL TRANSFER CANDIDATES ──────────────────────
print("── Generating transfer candidate pairs …")
print("   (This pairs every surplus dealer with every")
print("    urgent dealer for the same part)")

# Split into potential donors and receivers
donors    = dealer_part[dealer_part["Transferable_Units"] >= 1].copy()
receivers = dealer_part[dealer_part["Urgency_Score"]      >= 6].copy()

print(f"\n  Potential donor slots    : {len(donors):,}")
print(f"  Urgent receiver slots    : {len(receivers):,}")

# Cross-join on Part_Number (same part, different dealer)
candidates = donors.merge(
    receivers,
    on         = ["Part_Number","Description",
                  "Category_Group","Unit_Price"],
    suffixes   = ("_donor","_receiver"),
    how        = "inner"
)

# Remove self-transfers
candidates = candidates[
    candidates["Dealer_ID_donor"] !=
    candidates["Dealer_ID_receiver"]
].copy()

print(f"  Raw transfer pairs       : {len(candidates):,}")

# ── 9. COMPUTE FINAL TRANSFER UTILITY SCORE ──────────────────
print("── Computing Transfer Utility Scores …")

def compute_tus(row):
    friction = transfer_friction(
        row["Zone_donor"],
        row["Zone_receiver"],
        row["Unit_Price"],
        row["Transferable_Units_donor"]
    )
    tus = (row["Urgency_Score_receiver"]
           + row["Surplus_Score_donor"]
           + row["Criticality_Score_donor"]
           - friction)
    return round(min(max(tus, 0), 100), 2), round(friction, 2)

candidates[["TUS","Friction_Score"]] = candidates.apply(
    lambda r: pd.Series(compute_tus(r)), axis=1)

# Recommended transfer quantity = min(surplus, receiver needs)
candidates["Recv_Need"] = (
    candidates["Avg_Daily_Demand_receiver"] *
    candidates["Lead_Time_receiver"] * 1.5
).round(0)

candidates["Recommended_Qty"] = np.minimum(
    candidates["Transferable_Units_donor"],
    candidates["Recv_Need"]
).clip(lower=1).round(0)

# Financial impact
candidates["Transfer_Value"] = (
    candidates["Recommended_Qty"] *
    candidates["Unit_Price"]).round(0)

candidates["Holding_Cost_Relief"] = (
    candidates["Recommended_Qty"] *
    candidates["Unit_Price"] *
    HOLDING_RATE / 365 * 30      # monthly holding cost saved
).round(0)

# ── 10. FILTER & RANK ─────────────────────────────────────────
recommended = (candidates[candidates["TUS"] >= TRANSFER_THRESHOLD]
               .sort_values("TUS", ascending=False)
               .reset_index(drop=True))

print(f"\n  Pairs scoring ≥ {TRANSFER_THRESHOLD}        : "
      f"{len(recommended):,} recommended transfers")

# ── 11. SUMMARY TABLES ────────────────────────────────────────
print("\n" + "="*70)
print("  TRANSFER UTILITY SCORE — TOP 25 RECOMMENDED TRANSFERS")
print("="*70)

display_cols = [
    "Part_Number","Description","Category_Group",
    "Dealer_Name_donor","Zone_donor",
    "Dealer_Name_receiver","Zone_receiver",
    "Transferable_Units_donor","Recommended_Qty",
    "Days_Of_Stock_donor","Days_Of_Stock_receiver",
    "TUS","Urgency_Score_receiver","Surplus_Score_donor",
    "Criticality_Score_donor","Friction_Score",
    "Transfer_Value","Holding_Cost_Relief"
]

top25 = recommended.head(25)[display_cols]
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 200)

# Print in readable chunks
for _, row in top25.iterrows():
    print(f"\n  {'─'*66}")
    print(f"  Part     : {row['Description']} "
          f"({row['Part_Number']}) | {row['Category_Group']}")
    print(f"  FROM     : {row['Dealer_Name_donor']} "
          f"[{row['Zone_donor']}]  "
          f"→  Stock: {row['Transferable_Units_donor']:.0f} units  "
          f"| DOS: {row['Days_Of_Stock_donor']} days")
    print(f"  TO       : {row['Dealer_Name_receiver']} "
          f"[{row['Zone_receiver']}]  "
          f"→  DOS: {row['Days_Of_Stock_receiver']} days "
          f"(URGENT)")
    print(f"  MOVE     : {row['Recommended_Qty']:.0f} units  "
          f"| Value: ₹{row['Transfer_Value']:,.0f}  "
          f"| Holding relief: ₹{row['Holding_Cost_Relief']:,.0f}/mo")
    print(f"  SCORE    : TUS={row['TUS']:.1f}  "
          f"[Urgency={row['Urgency_Score_receiver']:.1f} + "
          f"Surplus={row['Surplus_Score_donor']:.1f} + "
          f"Criticality={row['Criticality_Score_donor']:.1f} - "
          f"Friction={row['Friction_Score']:.1f}]")

# ── 12. AGGREGATE INSIGHTS ────────────────────────────────────
print("\n\n" + "="*70)
print("  AGGREGATE TRANSFER INSIGHTS")
print("="*70)

print(f"\n── Total recommended transfers  : {len(recommended):,}")
print(f"── Total units to move          : "
      f"{recommended['Recommended_Qty'].sum():,.0f}")
print(f"── Total transfer value         : "
      f"₹{recommended['Transfer_Value'].sum():,.0f}")
print(f"── Total holding cost relief/mo : "
      f"₹{recommended['Holding_Cost_Relief'].sum():,.0f}")

print("\n── Transfers by Zone Direction:")
zone_flow = (recommended
    .groupby(["Zone_donor","Zone_receiver"])
    .agg(Transfers=("TUS","count"),
         Units    =("Recommended_Qty","sum"),
         Value    =("Transfer_Value", "sum"))
    .reset_index()
    .sort_values("Value", ascending=False))
print(zone_flow.to_string(index=False))

print("\n── Top Donor Dealers (most surplus to give):")
top_donors = (recommended
    .groupby("Dealer_Name_donor")
    .agg(Transfers=("TUS",             "count"),
         Units    =("Recommended_Qty", "sum"),
         Value    =("Transfer_Value",  "sum"),
         Avg_TUS  =("TUS",             "mean"))
    .sort_values("Value", ascending=False)
    .head(10))
print(top_donors.to_string())

print("\n── Top Receiver Dealers (most urgent need):")
top_receivers = (recommended
    .groupby("Dealer_Name_receiver")
    .agg(Transfers=("TUS",             "count"),
         Units    =("Recommended_Qty", "sum"),
         Value    =("Transfer_Value",  "sum"),
         Avg_TUS  =("TUS",             "mean"))
    .sort_values("Transfers", ascending=False)
    .head(10))
print(top_receivers.to_string())

print("\n── Transfers by Category:")
cat_summary = (recommended
    .groupby("Category_Group")
    .agg(Transfers=("TUS",             "count"),
         Units    =("Recommended_Qty", "sum"),
         Value    =("Transfer_Value",  "sum"),
         Avg_TUS  =("TUS",             "mean"))
    .sort_values("Value", ascending=False))
print(cat_summary.to_string())

# ── 13. SCORE DISTRIBUTION ANALYSIS ──────────────────────────
print("\n── TUS Score Distribution:")
bins = [0,20,40,55,70,85,101]
labels_b = ["0-20 No","20-40 Low","40-55 Marginal",
            "55-70 Recommended","70-85 Strong","85-100 Critical"]
candidates["TUS_Band"] = pd.cut(candidates["TUS"],
                                 bins=bins, labels=labels_b)
print(candidates["TUS_Band"].value_counts()
      .sort_index().to_string())

# ── 14. PLOTS ─────────────────────────────────────────────────
fig = plt.figure(figsize=(22, 18))
fig.suptitle("Heuristic Transfer Utility Score Dashboard",
             fontsize=17, fontweight="bold")
gs  = gridspec.GridSpec(3, 3, figure=fig,
                        hspace=0.55, wspace=0.4)

# Plot 1 — TUS score distribution
ax0 = fig.add_subplot(gs[0, :2])
ax0.hist(candidates["TUS"], bins=50,
         color="#3498DB", edgecolor="white", alpha=0.8)
ax0.axvline(TRANSFER_THRESHOLD, color="#E74C3C",
            linewidth=2.5, linestyle="--",
            label=f"Threshold = {TRANSFER_THRESHOLD}")
ax0.axvline(70, color="#F39C12",
            linewidth=1.5, linestyle=":",
            label="Strong (70)")
ax0.axvline(85, color="#2ECC71",
            linewidth=1.5, linestyle=":",
            label="Critical (85)")
ax0.set_title("Transfer Utility Score Distribution — All Pairs",
              fontweight="bold", fontsize=12)
ax0.set_xlabel("TUS Score (0–100)")
ax0.set_ylabel("Number of Candidate Pairs")
ax0.legend(); ax0.grid(alpha=0.3)

# Plot 2 — Component score breakdown (top 20 recommended)
ax1 = fig.add_subplot(gs[0, 2])
top20 = recommended.head(20)
bar_w = 0.6
ax1.barh(range(len(top20)),
         top20["Urgency_Score_receiver"],
         bar_w, label="Urgency",    color="#E74C3C", alpha=0.85)
ax1.barh(range(len(top20)),
         top20["Surplus_Score_donor"],
         bar_w,
         left=top20["Urgency_Score_receiver"],
         label="Surplus",    color="#3498DB", alpha=0.85)
ax1.barh(range(len(top20)),
         top20["Criticality_Score_donor"],
         bar_w,
         left=(top20["Urgency_Score_receiver"] +
               top20["Surplus_Score_donor"]),
         label="Criticality", color="#2ECC71", alpha=0.85)
ax1.barh(range(len(top20)),
         -top20["Friction_Score"],
         bar_w, label="Friction (–)",
         color="#E67E22", alpha=0.85)
ax1.set_title("TUS Components\n(Top 20 Transfers)",
              fontweight="bold", fontsize=12)
ax1.set_xlabel("Score")
ax1.legend(fontsize=8)
ax1.axvline(0, color="black", linewidth=0.8)

# Plot 3 — Zone flow heatmap
ax2 = fig.add_subplot(gs[1, 0])
zone_matrix = (recommended
    .groupby(["Zone_donor","Zone_receiver"])["Recommended_Qty"]
    .sum()
    .unstack(fill_value=0))
sns.heatmap(zone_matrix, annot=True, fmt=".0f",
            cmap="YlOrRd", ax=ax2,
            linewidths=0.5, linecolor="white",
            cbar_kws={"label":"Units"})
ax2.set_title("Zone-to-Zone\nTransfer Flow (Units)",
              fontweight="bold", fontsize=12)
ax2.set_xlabel("Receiver Zone")
ax2.set_ylabel("Donor Zone")

# Plot 4 — Transfer value by category
ax3 = fig.add_subplot(gs[1, 1])
cat_val = (recommended
    .groupby("Category_Group")["Transfer_Value"]
    .sum()
    .sort_values())
colors_c = ["#9B59B6","#2ECC71","#F39C12","#3498DB","#E74C3C"]
cat_val.plot(kind="barh", ax=ax3,
             color=colors_c[:len(cat_val)],
             edgecolor="white")
ax3.set_title("Transfer Value\nby Category",
              fontweight="bold", fontsize=12)
ax3.set_xlabel("Transfer Value (₹)")
ax3.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f"₹{x/1e6:.1f}M"))
ax3.grid(axis="x", alpha=0.3)

# Plot 5 — Top 15 receiver dealers by urgency
ax4 = fig.add_subplot(gs[1, 2])
recv_plot = (recommended
    .groupby("Dealer_Name_receiver")["TUS"]
    .mean()
    .sort_values(ascending=False)
    .head(15))
recv_plot.sort_values().plot(
    kind="barh", ax=ax4,
    color="#E74C3C", edgecolor="white", alpha=0.85)
ax4.set_title("Top 15 Receivers\nby Avg TUS",
              fontweight="bold", fontsize=12)
ax4.set_xlabel("Avg TUS Score")
ax4.grid(axis="x", alpha=0.3)

# Plot 6 — Scatter: Donor surplus vs Receiver urgency
# coloured by TUS
ax5 = fig.add_subplot(gs[2, :2])
sc  = ax5.scatter(
    recommended["Surplus_Score_donor"],
    recommended["Urgency_Score_receiver"],
    c=recommended["TUS"], cmap="RdYlGn",
    s=recommended["Transfer_Value"] / 5000,
    alpha=0.6, edgecolors="white", linewidth=0.3)
plt.colorbar(sc, ax=ax5, label="TUS Score")
ax5.set_title("Donor Surplus vs Receiver Urgency "
              "(size = Transfer Value, color = TUS)",
              fontweight="bold", fontsize=12)
ax5.set_xlabel("Donor Surplus Score (0–25)")
ax5.set_ylabel("Receiver Urgency Score (0–25)")
ax5.grid(alpha=0.3)
ax5.axvline(12, color="gray", linestyle="--",
            linewidth=1, alpha=0.5, label="Surplus midpoint")
ax5.axhline(12, color="gray", linestyle="--",
            linewidth=1, alpha=0.5, label="Urgency midpoint")
ax5.legend(fontsize=8)

# Plot 7 — Holding cost relief by donor dealer
ax6 = fig.add_subplot(gs[2, 2])
holding_plot = (recommended
    .groupby("Dealer_Name_donor")["Holding_Cost_Relief"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
    .sort_values())
holding_plot.plot(kind="barh", ax=ax6,
                  color="#9B59B6",
                  edgecolor="white", alpha=0.85)
ax6.set_title("Monthly Holding Cost\nRelief by Donor Dealer",
              fontweight="bold", fontsize=12)
ax6.set_xlabel("₹ Saved / Month")
ax6.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f"₹{x/1e3:.0f}K"))
ax6.grid(axis="x", alpha=0.3)

plt.tight_layout()
plt.savefig("transfer_utility_score.png",
            dpi=150, bbox_inches="tight")
plt.show()
print("\nSaved → transfer_utility_score.png")

# ── 15. EXPORT ────────────────────────────────────────────────
recommended[display_cols].to_csv(
    "transfer_recommendations.csv", index=False)
candidates[["Part_Number","Description","Category_Group",
            "Dealer_ID_donor","Dealer_Name_donor","Zone_donor",
            "Dealer_ID_receiver","Dealer_Name_receiver",
            "Zone_receiver","TUS","TUS_Band",
            "Recommended_Qty","Transfer_Value"]].to_csv(
    "transfer_all_candidates.csv", index=False)

print("Saved → transfer_recommendations.csv")
print("Saved → transfer_all_candidates.csv")
print(f"\n✅ Transfer Utility Score complete.")
print(f"   {len(recommended):,} transfers recommended above "
      f"threshold {TRANSFER_THRESHOLD}")
```
